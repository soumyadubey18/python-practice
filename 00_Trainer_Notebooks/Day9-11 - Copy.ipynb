# PART 1: SETS

## 1. What is a Set?

Think of a set like a **bag of unique marbles** - no two marbles can be the same color!

### Key Properties:

| Property | Description |
|----------|-------------|
| **Unordered** | Items have NO fixed position (no index!) |
| **Unique** | Duplicates are automatically removed |
| **Mutable** | Can add/remove items (but items themselves must be immutable) |
| **No Indexing** | Cannot use `set[0]` - will give error! |
| **Curly Braces** | Uses `{}` to create |

### When to Use Sets?
- Remove duplicates from a list
- Check if an item exists (very fast!)
- Find common/different items between groups
- Mathematical operations (union, intersection)

````python
# --- Creating Sets ---

# Method 1: Using curly braces
fruits = {"Apple", "Banana", "Mango", "Apple", "Banana"}
print("Fruits set:", fruits)        # Duplicates removed automatically!
print("Type:", type(fruits))

# Method 2: Empty set (IMPORTANT: {} creates a dict, NOT a set!)
empty_set = set()                   # Correct way!
wrong_way = {}                      # This is a DICTIONARY, not a set!
print("\nCorrect empty set:", empty_set, "Type:", type(empty_set))
print("Wrong way:", wrong_way, "Type:", type(wrong_way))

# Method 3: Convert list to set (removes duplicates!)
numbers_list = [1, 2, 3, 1, 2, 4, 5, 3]
numbers_set = set(numbers_list)
print("\nList:", numbers_list)
print("Set (duplicates gone!):", numbers_set)

# Note: True = 1 and False = 0 in sets
mixed = {1, True, 0, False, 2}
print("\nMixed set:", mixed)        # True and 1 are same, False and 0 are same
````

## 2. Adding & Removing Elements

Since sets are **mutable**, you can add/remove items. But you **CANNOT access by index**!

| Method | What it does |
|--------|-------------|
| `add(x)` | Add one item |
| `update([x,y])` | Add multiple items |
| `remove(x)` | Remove item (ERROR if not found) |
| `discard(x)` | Remove item (NO error if not found) |
| `pop()` | Remove & return a random item |
| `clear()` | Remove all items |

**Important:** Sets don't support indexing! `set[0]` will give `TypeError`

````python
# --- Adding & Removing Elements ---

colors = {"Red", "Green", "Blue"}
print("Original:", colors)

# ADD one item
colors.add("Yellow")
print("After add:", colors)

# ADD multiple items
colors.update(["White", "Black", "Orange"])
print("After update:", colors)

# REMOVE (raises error if not found)
colors.remove("Black")
print("After remove:", colors)

# DISCARD (no error if not found - safer!)
colors.discard("Purple")   # No error even though Purple doesn't exist
print("After discard:", colors)

# POP (removes random item - sets are unordered!)
popped = colors.pop()
print(f"Popped '{popped}':", colors)
````

## 3. Mathematical Set Operations

This is the **most powerful** feature of sets! Think of it like Venn diagrams.

```
Set A = {1, 2, 3, 4}          Set B = {3, 4, 5, 6}

    A only    Both    B only
    {1, 2}   {3, 4}   {5, 6}
```

| Operation | Symbol | Method | Result (A=\{1,2,3\}, B=\{3,4,5\}) |
|-----------|--------|--------|------|
| **Union** | `A \| B` | `A.union(B)` | {1,2,3,4,5} - ALL items |
| **Intersection** | `A & B` | `A.intersection(B)` | {3} - COMMON items |
| **Difference** | `A - B` | `A.difference(B)` | {1,2} - In A, NOT in B |
| **Symmetric Diff** | `A ^ B` | `A.symmetric_difference(B)` | {1,2,4,5} - NOT in both |
| **Subset** | `A <= B` | `A.issubset(B)` | Is A inside B? |
| **Superset** | `A >= B` | `A.issuperset(B)` | Does A contain B? |
| **Disjoint** | — | `A.isdisjoint(B)` | No common items? |

````python
# --- Mathematical Set Operations (Students in Clubs) ---

math_club = {"Rahul", "Priya", "Amit", "Sneha", "Kiran"}
science_club = {"Amit", "Sneha", "Deepak", "Meera", "Ravi"}

print("Math Club:", math_club)
print("Science Club:", science_club)

# UNION (|) - All students in either club
all_students = math_club.union(science_club)
print("\n1. Union (all students):", all_students)

# INTERSECTION (&) - Students in BOTH clubs
both_clubs = math_club.intersection(science_club)
print("2. Intersection (in both):", both_clubs)

# DIFFERENCE (-) - Only in math, NOT in science
only_math = math_club.difference(science_club)
print("3. Difference (only math):", only_math)

# SYMMETRIC DIFFERENCE (^) - In one club only, NOT both
one_club_only = math_club.symmetric_difference(science_club)
print("4. Symmetric Diff (one only):", one_club_only)

# SUBSET (<=) - Is a smaller set inside a bigger set?
toppers = {"Amit", "Sneha"}
print("\n5. Is toppers subset of math?", toppers.issubset(math_club))    # True
print("   Is math subset of toppers?", math_club.issubset(science_club))     # False

# SUPERSET (>=) - Does a bigger set contain a smaller set?
print("6. Is math superset of toppers?", math_club.issuperset(toppers) )  # True

# DISJOINT - No common elements at all?
sports_club = {"Vikas", "Raju", "Suman"}
print("7. Math & Sports disjoint?", math_club.isdisjoint(sports_club))  # True
print("   Math & Science disjoint?", math_club.isdisjoint(science_club))  # False
````

## 4. Frozenset (Immutable Set)

A **frozenset** is a set that **CANNOT be changed** after creation (like a tuple is to a list).

| Feature | `set` | `frozenset` |
|---------|-------|-------------|
| Mutable? | Yes | No |
| Can add/remove? | Yes | No |
| Use as dict key? | No | Yes |
| Use inside another set? | No | Yes |

````python
# --- Frozenset ---

# Create a frozenset
days = frozenset({"Mon", "Tue", "Wed", "Thu", "Fri"})
print("Frozenset:", days)
print("Type:", type(days))


# Frozenset supports set operations (read-only)
weekend = frozenset({"Sat", "Sun"})
print("\nAll days:", days | weekend)  # Union still works!
````

## 5. Practical Set Examples

````python
# --- Practical Example 1: Find Unique Vowels in a Word ---

word = "aeroplane"
vowels = {'a', 'e', 'i', 'o', 'u'}

# Convert word to set, then find common with vowels
word_letters = set(word)
unique_vowels = word_letters & vowels  # Intersection!

print(f"Word: '{word}'")
print(f"All unique letters: {word_letters}")
print(f"Unique vowels found: {unique_vowels}")

# --- Practical Example 2: Remove Duplicates from a List ---
print("\n" + "="*40)
names = ["Rahul", "Priya", "Amit", "Rahul", "Sneha", "Priya", "Amit"]
unique_names = list(set(names))   # Convert to set → back to list
print(f"Original list: {names}")
print(f"After removing duplicates: {unique_names}")

# --- Practical Example 3: Find Common Elements Between Two Lists ---
print("\n" + "="*40)
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
common = set(list1) & set(list2)
print(f"List 1: {list1}")
print(f"List 2: {list2}")
print(f"Common elements: {common}")
````

## 6. What is a Dictionary?

Think of a dictionary like a **phone book** - every name (key) has a number (value)!

### Key Properties:

| Property | Description |
|----------|-------------|
| **Key-Value Pairs** | Every item has a key and a value `{key: value}` |
| **Keys are Unique** | No duplicate keys (last value wins) |
| **Keys are Immutable** | Keys must be string, number, or tuple |
| **Values can be Anything** | Numbers, strings, lists, even other dicts |
| **Mutable** | Can add, change, remove items |
| **unOrdered** | Maintains insertion order (Python 3.7+) |
| **Curly Braces** | Uses `{}` with colon `:` syntax |

### When to Use Dictionary?
- Store related data together (like a student record)
- Fast lookup by key (much faster than searching a list)
- Count occurrences of items
- Group data by categories

````python
grocery = {"Rice": 80, "Wheat": 45, "Milk": 60, "Oil": 180, "Rice": 800}
grocery['Ragi']
grocery.get("Ragi")
grocery.get("Rice", "Not available")
````

````python
# --- Creating Dictionaries ---

# Empty dictionary
empty_dict = {}
print("Empty dict:", empty_dict, "Type:", type(empty_dict))

# Dictionary with data (like a grocery price list)
grocery = {"Rice": 80, "Wheat": 45, "Milk": 60, "Oil": 180, "Rice": 800}
print("Grocery:", grocery)

# Mixed value types (student record)
student = {
    "name": "Rahul",
    "age": 21,
    "branch": "CSE",
    "marks": [85, 90, 78],
    "is_active": True
}
print("Student:", student)

# IMPORTANT: Duplicate keys - last value wins!
prices = {"Apple": 100, "Banana": 40, "Apple": 150}
print("\nDuplicate keys:", prices)  # Apple will be 150 (last value wins)
````

## 7. Accessing Dictionary Values

Two ways to get values:

| Method | Syntax | If Key Missing |
|--------|--------|----------------|
| **Brackets** | `dict["key"]` | Raises `KeyError` |
| **get()** | `dict.get("key")` | Returns `None` (safe!) |
| **get() with default** | `dict.get("key", default)` | Returns the default value |

**Tip:** Always use `get()` when you're not sure if the key exists!

````python
# --- Accessing Dictionary Values ---

grocery = {"Rice": 80, "Wheat": 45, "Milk": 60, "Oil": 180}

# Method 1: Using brackets [] (raises error if key missing)
print("Rice price:", grocery["Rice"])         # 80
print("Oil price:", grocery["Oil"])           # 180

# Method 2: Using get() (safe - returns None if missing)
print("Milk price:", grocery.get("Milk"))     # 60
print("Sugar price:", grocery.get("Sugar"))   # None (no error!)

# Method 3: get() with default value
print("Sugar price:", grocery.get("Sugar", "Not available"))

# Check if key exists using 'in'
print("\nIs 'Rice' in grocery?", "Rice" in grocery)    # True
print("Is 'Sugar' in grocery?", "Sugar" in grocery)   # False
````

## 8. Modifying Dictionaries (Add, Update, Remove)

| Action | How to do it |
|--------|-------------|
| **Add new item** | `dict["new_key"] = new_value` |
| **Update existing** | `dict["key"] = new_value` |
| **Update multiple** | `dict.update({"k1": v1, "k2": v2})` |
| **Remove by key** | `dict.pop("key")` |
| **Remove last item** | `dict.popitem()` |
| **Delete a key** | `del dict["key"]` |
| **Clear all** | `dict.clear()` |

````python
student = {
    "name": "Rahul",
    "age": 21,
    "branch": "CSE",
    "marks": [85, 90, 78],
    "is_active": True
}
print("Student:", student)

student.pop('age')
print("Student:", student)

student.popitem()
print("Student:", student)

del student['name']
print("Student:", student)

student.clear()
print("Student:", student)
````

````python
# --- Modifying Dictionaries (Grocery Store Example) ---

grocery = {"Rice": 80, "Wheat": 45, "Milk": 60}
print("Original:", grocery)

# ADD a new item
grocery["Oil"] = 180
print("After adding Oil:", grocery)

# UPDATE an existing price
grocery["Rice"] = 95
print("After updating Rice:", grocery)

# UPDATE multiple items at once
grocery.update({"Milk": 65, "Sugar": 50, "Salt": 25})
print("After update multiple:", grocery)

# REMOVE using pop() (returns the removed value)
removed_price = grocery.pop("Salt")
print(f"Removed Salt (price was {removed_price}):", grocery)

# REMOVE last item using popitem()
last_item = grocery.popitem()
print(f"Removed last item {last_item}:", grocery)

# DELETE using del
del grocery["Wheat"]
print("After del Wheat:", grocery)
````

## 9. Dictionary Methods

| Method | What it does | Returns |
|--------|-------------|---------|
| `keys()` | All keys | dict_keys object |
| `values()` | All values | dict_values object |
| `items()` | All key-value pairs | dict_items (tuples) |
| `get(key, default)` | Get value safely | value or default |
| `pop(key)` | Remove & return value | removed value |
| `popitem()` | Remove last pair | (key, value) tuple |
| `update(dict2)` | Add/update from another dict | None |
| `clear()` | Remove all items | None |
| `fromkeys(keys, value)` | Create dict with default value | new dict |

````python
# --- Dictionary Methods ---

fruits = {"Apple": 102, "Orange": 300, "Banana": 45, "Mango": 200}

# keys() - Get all keys
print("Keys:", list(fruits.keys()))

# values() - Get all values
print("Values:", list(fruits.values()))

# items() - Get all key-value pairs as tuples
print("Items:", list(fruits.items()))

# fromkeys() - Create dict with default value
subjects = dict.fromkeys(["Math", "Science", "English"], 0)
print("\nfromkeys:", subjects)  # All subjects with score 0
````

## 10. Looping Through Dictionaries

Three ways to loop:
```python
for key in dict:              # Loop through keys only
for key, value in dict.items():  # Loop through both (most common)
for value in dict.values():      # Loop through values only
```

````python
# --- Looping Through Dictionaries ---

student = {"name": "Priya", "age": 22, "branch": "ECE", "city": "Chennai"}

# Method 1: Loop through keys only
print("Keys:")
for keys in student:
    print(keys, student[keys])
````

## 12. Quick Comparison: List vs Tuple vs Set vs Dictionary

| Feature | List `[]` | Tuple `()` | Set `{}` | Dict `{k:v}` |
|---------|-----------|-----------|----------|--------------|
| Ordered? | Yes | Yes | No | No |
| Mutable? | Yes | No | Yes | Yes |
| Duplicates? | Yes | Yes | No | Keys: No |
| Indexing? | Yes | Yes | No | By key |
| Use case | General collection | Fixed data | Unique items | Key-value pairs |

```python
my_list  = [1, 2, 3, 3]       # Ordered, duplicates OK
my_tuple = (1, 2, 3, 3)       # Ordered, cannot change
my_set   = {1, 2, 3, 3}       # Unordered, no duplicates → {1, 2, 3}
my_dict  = {"a": 1, "b": 2}   # Key-value pairs
```
