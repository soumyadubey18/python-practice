## 1. What is a Function?

A function is a **reusable block of code** that performs a specific task.

**Analogy:** Think of a function like a **juice machine** — you put fruits in (input), it processes them, and gives you juice (output). You can reuse the same machine for different fruits!

### Why Use Functions?
- **Reusability** — Write once, use many times
- **Organization** — Break big problems into small pieces
- **Easy Debugging** — Fix one place, works everywhere

### Syntax:
```python
def function_name(parameters):    # define
    # code block
    return result                  # optional

function_name(arguments)           # call
```

### Types of Functions:

| Type | Description |
|------|-------------|
| **Built-in** | Already available — `print()`, `len()`, `max()`, `sum()` |
| **User Defined (UDF)** | Created by you using `def` keyword |
| **Lambda** | One-line anonymous functions |
| **Recursive** | A function that calls itself |

## 2. Creating Functions

### Two patterns:

| Pattern | When to use | Example |
|---------|------------|---------|
| **Without return** (prints result) | Display output directly | `print(f"Sum = {a+b}")` |
| **With return** (gives back value) | Store/use result later | `return a + b` |

**Key difference:**
- `print()` → shows on screen, value is lost
- `return` → sends value back, you can store it in a variable

````python
# --- Without return (just prints) ---
def greet(name):
    print(f"Hello {name}, welcome to Python!")

greet("Ravi")
greet("Sneha")

print()

# --- With return (gives value back) ---
def add(a, b):
    return a + b

result = add(10, 20)          # result stores 30
print("Sum:", result)
print("Double:", add(5, 5) * 2)  # can use returned value in expressions
````

````python
def add_num(a, b):
   print(a+b)
   print(a-b)
   print(a/b)
   print("*"*10)

add_num(10,20)
add_num(345,567)
add_num(200,567)
````

````python
def paswword_validator(pwd):

    upper_count = 0
    lower_count = 0
    sp_count = 0
    space = 0

    if len(pwd) < 8:
        print("password should be atleast 8 characters")
    else:
        for char in pwd:
            if char.isupper():
                upper_count +=1
            if char.islower():
                lower_count +=1
            if not char.isalnum() and not char.isspace():
                sp_count += 1
            if char.isspace():
                space += 1

    if lower_count >= 1 and upper_count >= 1 and sp_count >= 1 and space == 0:
        print("valid password")
    else:
        print("invalid password")

paswword_validator("Maheshkumar@123")
paswword_validator("Maheshkumar123")
paswword_validator("Kumar#1")

````

## 3. Function Arguments

### 5 Types of Arguments:

| Type | Syntax | Key Rule |
|------|--------|----------|
| **Positional** | `def func(a, b)` → `func(10, 20)` | Order matters! |
| **Keyword** | `def func(a, b)` → `func(b=20, a=10)` | Order doesn't matter |
| **Default** | `def func(a, b=10)` → `func(5)` | Uses default if not provided |
| ***args** | `def func(*args)` → `func(1,2,3,4)` | Accepts any number of positional args (as tuple) |
| ****kwargs** | `def func(**kwargs)` → `func(x=1, y=2)` | Accepts any number of keyword args (as dict) |

### Rules:
- Positional args must come **before** keyword args
- Default args must come **after** non-default args
- Order in definition: `def func(positional, *args, **kwargs)`

````python
# --- Positional Arguments (order matters!) ---
def power(base, exp):
    return base ** exp

print(power(2, 10))    # 2^10 = 1024
print(power(10, 2))    # 10^2 = 100  (different result, same numbers!)
````

````python
# --- Keyword Arguments (order doesn't matter) ---
def power(base, exp):
    return base ** exp

print(power(base= 2, exp=10))  # 2^10 = 1024 (same as first call)
print(power(exp=10, base= 2))  # 2^10 = 1024 (same as first call)
````

````python
def power(base):
    return base

print(power()) 
print(power(10)) 
````

````python

# --- Default Arguments ---
def welcome(name, course="Python"):
    print(f"Hi {name}, welcome to {course} class!")

welcome("Ravi")                  # uses default: Python
welcome("Sneha", "Data Science") # overrides default
````

````python
# --- *args (variable length arguments) (variable number of positional arguments) ---
def add_all(*args):
    print(f"Received: {args}")   # tuple
    print(f"Sum: {sum(args)}")

add_all()
add_all(10, 20)
add_all(1, 2, 3, 4, 5)
````

````python
def power(**base):
    return base

print(power(a= 10)) 
````

````python


# --- **kwargs (variable number of keyword arguments) ---
def student_info(**kwargs):
    print(f"Received: {kwargs}")   # dictionary

student_info(name="Ravi", age=25, city="Hyderabad")
student_info(name="Ravi" )
student_info()
````

````python
def demo(a, b, *args, **kwargs):
    print(f"a={a}, b={b}, args={args}, kwargs={kwargs}")

demo(1, 30)
demo(1, 30, 10, 20)
demo(1, 30, 10, 20, x=100)
````

````python
# --- Default Arguments ---
def welcome(name = "Your name"):
    print(f"Hi {name}, welcome to python class!")

welcome("Ravi")
welcome()
````

## 4. Lambda Functions

A **lambda** is a small, one-line, anonymous (nameless) function.

### When to use?
- For simple operations you don't want to write a full `def` for
- Often used with `map()`, `filter()`, `sorted()`

### Syntax:
```python
lambda arguments: expression
```

### Comparison:
```python
# Regular function          # Lambda equivalent
def add(a, b):              add = lambda a, b: a + b
    return a + b
```

````python
def calc_power(a,b):
    return a**b

calc_power(10,20)
````

````python
(lambda a,b : a**b)(10,20)
(lambda a,b : a**b)(34,20)
````

````python
Lam_pow =(lambda a,b: a**b)
print(Lam_pow(10,20))
print(Lam_pow(100,20))
````

````python
indian_names = ["Aarav","Rahul","Priya","Ananya","Vikram","Sneha","Arjun","Kavya","Rohan","Neha"]

res = map(lambda x: len(x), indian_names)
print(res)
print(list(res))


````

## 5. Recursive Functions

A recursive function **calls itself** to break a big problem into smaller identical problems.

### Two requirements:
1. **Base case** — when to STOP (prevents infinite loop)
2. **Recursive case** — call itself with a smaller problem

### Example: Factorial (5! = 5 × 4 × 3 × 2 × 1 = 120)

```
factorial(5)
    → 5 * factorial(4)
        → 4 * factorial(3)
            → 3 * factorial(2)
                → 2 * factorial(1)
                    → 1 * factorial(0)
                        → returns 1  ← BASE CASE
                    → returns 1 * 1 = 1
                → returns 2 * 1 = 2
            → returns 3 * 2 = 6
        → returns 4 * 6 = 24
    → returns 5 * 24 = 120
```

````python
# without recursion 

def factorial(n):
    fact = 1 
    if n == 0:
        return 1
    else:
        while n > 0:
            fact = fact * n
            n = n - 1
    return fact
    
print(factorial(0))
````

````python
def factorial(n):
    fact = 1 
    if n == 0:
        return 1
    else:  
        return n * factorial(n-1)
    
print(factorial(5))
````

````python
# --- Recursive Function: Factorial ---
def factorial(n):
    print("n = ", n)
    if n == 0:            
        return 1
    else:
                # recursive case
        return n * factorial(n - 1)

print("5! =", factorial(5))    # 120
# print("3! =", factorial(3))    # 6
# print("0! =", factorial(0))    # 1
````

## 6. Practice Problem — Password Validator

**Rules:**
- At least 8 characters
- At least 1 uppercase letter
- At least 1 lowercase letter
- At least 1 digit
- At least 1 special symbol
- No spaces allowed

````python
# --- Password Validator Function ---
def validate_password(password):
    special_chars = [chr(i) for i in range(33, 127) if not chr(i).isalnum()]

    upper = lower = digit = special = space = 0

    for char in password:
        if char.isupper():    upper += 1
        elif char.islower():  lower += 1
        elif char.isdigit():  digit += 1
        elif char.isspace():  space += 1
        elif char in special_chars: special += 1

    # Check all conditions
    if len(password) >= 8 and upper >= 1 and lower >= 1 and digit >= 1 and special >= 1 and space == 0:
        print(f"'{password}' → VALID")
    else:
        print(f"'{password}' → INVALID")
        if len(password) < 8:  print("  - Min 8 characters needed")
        if upper == 0:         print("  - Need at least 1 uppercase")
        if lower == 0:         print("  - Need at least 1 lowercase")
        if digit == 0:         print("  - Need at least 1 digit")
        if special == 0:       print("  - Need at least 1 special char")
        if space > 0:          print("  - No spaces allowed")

validate_password("Python@123")
validate_password("weak")
validate_password("NoDigit!Here")
````

## Summary

| Concept | Key Point |
|---------|-----------|
| `def` function | Reusable code block, can have parameters and return values |
| `return` vs `print` | `return` gives value back; `print` just displays |
| Positional args | Order matters: `func(10, 20)` |
| Keyword args | Order doesn't matter: `func(b=20, a=10)` |
| Default args | Fallback value: `def func(x=10)` |
| `*args` | Collects extra positional args as **tuple** |
| `**kwargs` | Collects extra keyword args as **dictionary** |
| Lambda | One-line function: `lambda x: x * 2` |
| Recursion | Function calls itself; needs a base case to stop |
