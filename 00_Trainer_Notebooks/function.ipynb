```python
def remove_duplicates(nums):

    unique_nums = []
    for num in nums:
        if num not in unique_nums:
            unique_nums.append(num)

    return unique_nums
    

nums = [1,12,23,4,54,2,4,5,6,8,5,6,8,5]
remove_duplicates(nums)
```

```python
lst = [5,6,7,8,1,5]

target = 12

def sum_of_two(lst,target):
    first = lst[0]
    for num in range(1,len(lst)):
        if target == first + lst[num]:
            print(first,lst[num])

sum_of_two(lst, target)
```

```python
#recursive function...

def factorial(n):
    result = 1
    if n <=0:
        return result
    else:
        print("n = ",n)
        result = n * factorial(n-1)
        print("Result = ",result)
        return result
factorial(3)
```
