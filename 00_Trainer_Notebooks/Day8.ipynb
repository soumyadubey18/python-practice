## 1. What is a Tuple?

Think of a tuple like a **sealed box** - once you pack items in it, you **cannot change** them!

- A tuple stores multiple values in one variable
- Uses parentheses `()` for representation
- Items are separated by commas
- Can hold numbers, text, or a mix of both
- Items have a position (index) starting from **0**
- **Immutable** - you CANNOT add, remove, or change items after creation

**When to use Tuple instead of List?**
- When data should NOT be changed (e.g., days of the week, coordinates)
- Tuples are **faster** than lists
- Can be used as dictionary keys (lists cannot)

````python
# Empty tuple
empty_tup = () 
print("Empty tuple:", empty_tup, "| Type:", type(empty_tup))
````

## 2. Accessing Elements (Indexing & Slicing)

Works exactly like lists:
- **Positive index**: Count from left (starts at 0)
- **Negative index**: Count from right (starts at -1)
- **Slicing**: `tuple[start : stop : step]`

```
Items:    "Mon"  "Tue"  "Wed"  "Thu"  "Fri"
Index:      0      1      2      3      4
Neg Index: -5     -4     -3     -2     -1
```

````python
# --- Accessing Tuple Elements ---

colors = ("Red", "Green", "Blue", "Yellow", "White")

# Positive indexing
print("First color:", colors[0])       # Red
print("Third color:", colors[2])       # Blue

# Negative indexing
print("Last color:", colors[-1])       # White
print("Second last:", colors[-2])      # Yellow

# Slicing: tuple[start : stop : step]
print("First 3:", colors[0:3])         # ('Red', 'Green', 'Blue')
print("From index 2:", colors[2:])     # ('Blue', 'Yellow', 'White')
print("Reverse:", colors[::-1])        # Reversed tuple
print("Skip one:", colors[::2])        # ('Red', 'Blue', 'White')
````

## 3. Tuples are Immutable (Cannot Change!)

The **biggest difference** between a tuple and a list:
- **List** → You CAN change items (mutable)
- **Tuple** → You CANNOT change items (immutable)

Trying to modify a tuple gives a `TypeError`!

````python
# --- Tuple is Immutable ---

fruits = ("Apple", "Banana", "Mango","Apple")
print("Original tuple:", fruits)

# Try to change an item → ERROR!

# fruits[0] = "Orange"

# Try to add an item → ERROR!
# fruits.append("Grapes")

fruits.count("Apple")

fruits.index("Apple")


# WORKAROUND: Convert to list, modify, convert back
fruits_list = list(fruits)       # Convert tuple → list
fruits_list.append("Grapes")     # Modify the list
fruits = tuple(fruits_list)      # Convert list → tuple
print("After workaround:", fruits)
````

## 4. Tuple Operations

| Operation | Symbol | Example | Result |
|-----------|--------|---------|--------|
| Concatenation | `+` | `(1,2) + (3,4)` | `(1,2,3,4)` |
| Repetition | `*` | `("Hi",) * 3` | `("Hi","Hi","Hi")` |
| Membership | `in` | `3 in (1,2,3)` | `True` |
| Length | `len()` | `len((1,2,3))` | `3` |

````python
# --- Tuple Operations ---

# Concatenation: Joining two tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined = tuple1 + tuple2
print("Concatenation:", combined)       # (1, 2, 3, 4, 5, 6)

# Repetition: Repeating a tuple
greeting = ("Hello!",) * 3
print("Repetition:", greeting)          # ('Hello!', 'Hello!', 'Hello!')

# Membership: Check if item exists
numbers = (10, 20, 30, 40, 50)
print("Is 30 in tuple?", 30 in numbers)     # True
print("Is 99 not in tuple?", 99 not in numbers)     # True

# Length: Count total items
print("Total items:", len(numbers))          # 5
````

## 5. Tuple Unpacking

**Unpacking** means assigning each tuple item to a separate variable in one line.

```python
# Instead of:
name = student[0]
age = student[1]

# You can do:
name, age = student   # This is unpacking!
```

````python
# --- Tuple Unpacking ---

# Basic unpacking
student = ("Priya", 22, "ECE")
name, age, branch = student
name, age, branch = ("Priya", 22, "ECE")

print(f"Name: {name}, Age: {age}, Branch: {branch}")
````

````python
# Swapping two variables 
a = 10
b = 20

# swap two variables using temp variable
temp = a
a = b
b = temp

print(f"Before swap: a={a}, b={b}")
a, b = b, a  
print(f"After swap: a={a}, b={b}")
````

## 6. Tuple Methods & Built-in Functions

Tuples have only **2 methods** (because they are immutable):

| Method/Function | What it does |
|----------------|-------------|
| `tuple.count(x)` | Counts how many times `x` appears |
| `tuple.index(x)` | Returns the index of first occurrence of `x` |
| `len(tuple)` | Total number of items |
| `max(tuple)` | Largest item |
| `min(tuple)` | Smallest item |
| `sum(tuple)` | Sum of all items (numbers only) |
| `sorted(tuple)` | Returns a sorted **list** (not tuple) |

````python
# --- Tuple Methods & Functions ---

marks = (78, 92, 85, 92, 67, 92, 45)

# count() - How many times does 92 appear?
print("Count of 92:", marks.count(92))       # 3

# index() - At which position is 85?
print("Index of 85:", marks.index(85))       # 2

# Built-in functions
print("Length:", len(marks))         # 7
print("Maximum:", max(marks))        # 92
print("Minimum:", min(marks))        # 45
print("Sum:", sum(marks))            # 451

# sorted() returns a LIST, not a tuple!
sorted_marks = sorted(marks)
print("Sorted:", sorted_marks)               # [45, 67, 78, 85, 92, 92, 92]
print("Type of sorted:", type(sorted_marks)) # <class 'list'>
````

## 7. Looping Through Tuples

Two common ways to loop:
- `for item in tuple` → Get values only
- `for index, item in enumerate(tuple)` → Get both index and value

````python
# --- Looping Through Tuples ---

sports = ("Cricket", "Football", "Hockey", "Tennis", "Badminton")

# Method 1: Simple for loop (value only)
print("All sports:")
for sport in sports:
    print(f"  - {sport}")

# Method 2: Using enumerate (index + value)
print("\nWith index:")
for idx, sport in enumerate(sports):
    print(f"  {idx}. {sport}")

# Method 3: Using range and len
print("\nUsing range:")
for i in range(len(sports)):
    print(f"  Position {i}: {sports[i]}")
````

## 8. Tuple vs List - Quick Comparison

| Feature | List `[]` | Tuple `()` |
|---------|-----------|------------|
| Mutable? | Yes (can change) | No (cannot change) |
| Speed | Slower | Faster |
| Methods | Many (append, remove, sort...) | Only 2 (count, index) |
| Use case | Data that changes | Data that stays fixed |
| array | Dynamic array | static array |
| Memory | More | Less |

**Rule of thumb:** If data should NOT change → use Tuple. If it might change → use List.

````python
# --- Tuple vs List Demo ---

# List - mutable (can change)
my_list = [1, 2, 3]
my_list[0] = 100        # Works fine!
my_list.append(4)       # Works fine!
print("List after changes:", my_list)

# Tuple - immutable (cannot change)
my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 100   # ERROR!
except TypeError as e:
    print(f"Tuple error: {e}")

# Tuple is faster (use for fixed data)
import sys
print(f"\nMemory: List={sys.getsizeof(my_list)} bytes, Tuple={sys.getsizeof(my_tuple)} bytes")
````
