## 1. Mutable vs Immutable

### What does it mean?
- **Immutable** = CANNOT change after creation → creates a NEW object (new `id()`)
- **Mutable** = CAN change in place → same object (same `id()`)

| Immutable (cannot change) | Mutable (can change) |
|---------------------------|---------------------|
| `int`, `float`, `str` | `list`, `dict`, `set` |
| `tuple`, `frozenset`, `bool` | `bytearray` |

**How to verify?** Use `id()` — if the id changes, a new object was created (immutable behavior).

````python
# --- Immutable: String (id changes when value changes) ---
name = "Mahesh"
print(f"name = '{name}', id = {id(name)}")

name =  name + " Kumar"
print(f"name = '{name}', id = {id(name)}")  # Different id! New object created.

print()

# --- Mutable: List (id stays same even after modification) ---
grocery = ['Rice', 'Oil', 'Sugar']
print(f"grocery = {grocery}, id = {id(grocery)}")

grocery.append('Atta')
print(f"grocery = {grocery}, id = {id(grocery)}")  # Same id! Modified in place.
````

## 2. input() Function

`input()` takes user input from keyboard. **Always returns a string!**

```python
data = input("Enter something: ")   # data is always str
```

**To get a number:** Convert using `int()` or `float()`
```python
num = int(input("Enter a number: "))
```

````python
# --- input() always returns string ---
data = input("Enter your name: ")
print(data, "→ type:", type(data))

# --- Type conversion for numbers ---
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))
print(f"{num1} + {num2} = {num1 + num2}")
````

## 3. print() Function

### Syntax:
```python
print(value1, value2, sep=" ", end="\n")
```

| Parameter | Default | Purpose |
|-----------|---------|---------|
| `sep` | `" "` (space) | Separator between multiple values |
| `end` | `"\n"` (newline) | What to print at the end |

````python
# --- print() with sep and end ---

# Default: space separator, newline at end
print("Hello", "World")          # Hello World

# Custom separator
day = 10
month = 10
year = 2026
print(day, month, year)    # 10 5 1996
print(day, month, year, sep="-")    # 10/5/1996

````

````python
# Custom end (no newline)
print("Hello", end=" ")
print("World")                      # Hello World (on same line)

# Combine both
print("A", "B", "C", sep=" → ", end=" ✓\n")  # A → B → C ✓
````

## 4. == vs is (Value vs Identity)

| Operator | Checks | Example |
|----------|--------|---------|
| `==` | Are the **values** equal? | `[1,2] == [1,2]` → `True` |
| `is` | Are they the **same object** in memory? | `[1,2] is [1,2]` → `False` |
| `is not` | Are they **different objects**? | `[1,2] is not [1,2]` → `True` |

**Rule:** `==` compares content. `is` compares memory address (`id()`).

````python
# --- == vs is ---

# Small integers: Python caches them (same object)
a = 10
c = 10
print(f"a == c: {a == c}")    # True (same value)
print(f"a is c: {a is c}")    # True (same object — Python optimizes small ints)
print(f"id(a)={id(a)}, id(c)={id(c)}")

print()

# Lists: Even with same content, they are different objects
lst1 = [1, 2, 3]
lst2 = [1, 2, 3]
print(f"lst1 == lst2: {lst1 == lst2}")    # True (same content)
print(f"lst1 is lst2: {lst1 is lst2}")    # False (different objects!)
print(f"id(lst1)={id(lst1)}, id(lst2)={id(lst2)}")
````
