## 1. What is a String?

A string is a **sequence of characters** enclosed in quotes - like text in a message!

### Key Properties:

| Property | Description |
|----------|-------------|
| **Ordered** | Characters have a fixed position (index) |
| **Immutable** | Cannot change individual characters |
| **Iterable** | Can loop through each character |
| **Indexed** | Access characters using `[index]` |

### 4 Ways to Create Strings:
```python
'single quotes'       → For simple strings
"double quotes"       → For strings with apostrophe
'''triple single'''   → For multiline strings
"""triple double"""   → For multiline strings
```

````python
print('single quotes')      
print("double quotes")     
print('''triple single''')   
print("""triple double""")
````

````python
# --- Creating Strings ---

# Single quotes
name = 'Python'
print(name, "| Type:", type(name))

# Double quotes (useful when string has apostrophe)
sentence = "I can't stop learning!"
print(sentence)


# sentence = "I can't 
# stop 
# learning!"
# print(sentence)

# Triple quotes (for multiline text)
address = """123, MG Road,
Hyderabad,
Telangana - 500001"""
print(address)

````

## 2. String Indexing & Slicing

Strings work like lists - each character has a position number!

```
String:    P    y    t    h    o    n 
Index:     0    1    2    3    4    5
Neg Index: -6   -5   -4   -3   -2   -1
```

**Slicing syntax:** `string[start : stop : step]`

````python
# --- String Indexing & Slicing ---

name = "Python"

# Indexing (single character)
print("First char:", name[0])        # P
print("Last char:", name[-1])        # n
print("Third char:", name[2])        # t

# Slicing (substring)
print("First 3:", name[0:3])         # Pyt
print("From index 2:", name[2:])     # thon
print("Last 3:", name[-3:])          # hon
print("Skip one:", name[::2])        # Pto
print("Reverse:", name[::-1])        # nohtyP

# Length
print("Length:", len(name))          # 6

# Loop through characters
for char in name:
    print(char, end=" ")
print()
````

## 3. Strings are Immutable!

You **CANNOT** change a character in a string directly. You must create a new string.

```python
name = "Python"
name[0] = "p"   # ERROR! TypeError!
```

**Workaround:** Use string methods (they return NEW strings)

````python
# --- Strings are Immutable ---

name = "abbabababababab"

# name[0] = "p"

# Using replace() method - returns NEW string
result = name.replace("a", "A")

print("Original still:", name)  # abbabababababab (unchanged!)
print("Replaced:", result)   # AbbAbAbAbAbAbAb
````

````python
# atlest one Uppercase alphabet
# atlest one lowercase alphabet
# atlest one special character
# no space allowed
# minimum 8 characters should be there

pwd = "maheshkumar@123"

upper_count = 0
lower_count = 0
sp_count = 0
space = 0

if len(pwd) < 8:
    print("password should be atleast 8 characters")
else:
    for char in pwd:
        if char.isupper():
            upper_count +=1
        if char.islower():
            lower_count +=1
        if not char.isalnum() and not char.isspace():
            sp_count += 1
        if char.isspace():
            space += 1

if lower_count >= 1 and upper_count >= 1 and sp_count >= 1 and space == 0:
    print("valid password")
else:
    print("invalid password")

if all([lower_count,upper_count,sp_count]) and space == 0:
    print("valid password")
else:
    print("invalid password")

````

## 5. Split, Join & Strip

| Method | What it does | Example |
|--------|-------------|---------|
| `split(sep)` | Break string into list | "a,b,c".split(",") → ['a','b','c'] |
| `join(list)` | Join list into string | "-".join(['a','b']) → "a-b" |
| `strip()` | Remove spaces from both ends | "  hi  ".strip() → "hi" |
| `lstrip()` | Remove spaces from left | "  hi  ".lstrip() → "hi  " |
| `rstrip()` | Remove spaces from right | "  hi  ".rstrip() → "  hi" |

````python
date = "25/12/2024"
print(date)
res = date.split(sep="/")

print(res)

result = "-".join([res[1],res[0],res[2]])
result
````

## 6. Search Methods (find, index)

| Method | What it does | Not found? |
|--------|-------------|------------|
| `find(sub)` | First occurrence index (left→right) | Returns **-1** |
| `rfind(sub)` | First occurrence index (right→left) | Returns **-1** |
| `index(sub)` | Same as find | Raises **ValueError** |
| `rindex(sub)` | Same as rfind | Raises **ValueError** |

**Tip:** Use `find()` to safely check if a substring exists. Use `index()` only when you're sure it exists.

````python
# --- Search Methods ---

text = "Python programming is fun. Python is powerful."

# find() - returns index, or -1 if not found
print("find('Python'):", text.find("Python"))       # 0 (first occurrence)
print("find('Java'):", text.find("Java"))           # -1 (not found)

# rfind() - search from right
print("rfind('Python'):", text.rfind("Python"))     # 27 (last occurrence)

# index() - same as find but raises error if not found
print("index('fun'):", text.index("fun"))           # 23

# Practical: Check if email is valid format
email = "user@gmail.com"
if email.find("@") != -1 and email.find(".") != -1:
    print(f"'{email}' looks like a valid email")
else:
    print(f"'{email}' is not a valid email")
````

3 ways to insert variables into strings:

| Method | Syntax | Best For |
|--------|--------|----------|
| **f-string** (Python 3.6+) | `f"Hello {name}"` | Modern, clean, fast |
| **format()** | `"Hello {}".format(name)` | Older code |
| **Concatenation** | `"Hello " + name` | Simple joins |

````python
# --- String Formatting ---

name = "Ravi"
age = 25
price = 99.5

# Method 1: f-string (RECOMMENDED)
print(f"My name is {name} and I am {age} years old")

# Method 2: format()
print("My name is {} and I am {} years old".format(name, age))

# Method 3: Concatenation (only for strings!)
print("My name is " + name +  " and I am" + " 25 " + "years old")
````

## 8. Escape Characters & ASCII

### Escape Characters:
| Code | What it does |
|------|-------------|
| `\'` | Single quote inside single-quoted string |
| `\\` | Backslash character |
| `\n` | New line |
| `\t` | Tab (8 spaces) |

### ASCII (American Standard Code for Information Interchange):
| Range | Characters |
|-------|-----------|
| 65 - 90 | A to Z (uppercase) |
| 97 - 122 | a to z (lowercase) |
| 48 - 57 | 0 to 9 (digits) |

**Functions:** `ord('A')` → 65 (character → number) | `chr(65)` → 'A' (number → character)

````python
# --- Escape Characters ---

# Newline \n
print("Line 1\nLine 2\nLine 3")

# Tab \t
print("Name\tAge\tCity")
print("Ravi\t25\tHyderabad")

# Backslash \\
print("File path: C:\\Users\\Documents")

# Quote inside string
print('It\'s a beautiful day')

print()

# --- ASCII Operations ---

# ord() → character to number
print("ord('A'):", ord('A'))   # 65
print("ord('a'):", ord('a'))   # 97
print("ord('0'):", ord('0'))   # 48

# chr() → number to character
print("chr(65):", chr(65))     # A
print("chr(97):", chr(97))     # a
print("chr(48):", chr(48))     # 0

# Print all uppercase letters using ASCII
print("A-Z:", end=" ")
for i in range(65, 91):
    print(chr(i), end=" ")
print()
````

## 9. String Validation Methods

These methods return **True** or **False** — useful for checking user input!

| Method | Returns True if... |
|--------|-------------------|
| `isalpha()` | Only letters (no spaces, numbers) |
| `isdigit()` | Only digits (0-9) |
| `isalnum()` | Letters OR digits (no special chars) |
| `isupper()` | All letters are uppercase |
| `islower()` | All letters are lowercase |
| `isspace()` | Only whitespace characters |
| `istitle()` | Each word starts with capital |

````python
# --- String Validation Methods ---

# isalpha() - only letters?
print("'Hello'.isalpha():", "Hello".isalpha())       # True
print("'Hello World'.isalpha():", "Hello World".isalpha())  # False (space!)

# isdigit() - only digits?
print("'12345'.isdigit():", "12345".isdigit())       # True
print("'123.45'.isdigit():", "123.45".isdigit())     # False (dot!)

# isalnum() - letters or digits?
print("'Python3'.isalnum():", "Python3".isalnum())   # True
print("'user@1'.isalnum():", "user@1".isalnum())     # False (@ symbol!)

# isupper() / islower()
print("'HELLO'.isupper():", "HELLO".isupper())       # True
print("'hello'.islower():", "hello".islower())       # True

print()

# Practical: Simple input validation
username = "Ravi123"
if username.isalnum():
    print(f"'{username}' is valid (only letters & numbers)")
else:
    print(f"'{username}' is invalid (has special characters)")
````

````python
# --- Practice Problem 1: Palindrome Checker ---
# A palindrome reads same forward and backward (madam, racecar)

word = "racecar"
reversed_word = word[::-1]

if word == reversed_word:
    print(f"'{word}' is a Palindrome!")
else:
    print(f"'{word}' is NOT a Palindrome")

# --- Practice Problem 2: Password Validator ---
# Rules: At least 8 chars, has uppercase, lowercase, and digit

password = "Python@123"
errors = []

if len(password) < 8:
    errors.append("Must be at least 8 characters")
if password == password.lower():
    errors.append("Must have at least 1 uppercase letter")
if password == password.upper():
    errors.append("Must have at least 1 lowercase letter")

has_digit = False
for ch in password:
    if ch.isdigit():
        has_digit = True
        break
if not has_digit:
    errors.append("Must have at least 1 digit")

if errors:
    print(f"Password '{password}' is WEAK:")
    for err in errors:
        print(f"  - {err}")
else:
    print(f"Password '{password}' is STRONG!")

# --- Practice Problem 3: Character Counter ---
text = "Hello World"
letters = digits = spaces = special = 0

for ch in text:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1

print(f"\nText: '{text}'")
print(f"Letters: {letters}, Digits: {digits}, Spaces: {spaces}, Special: {special}")
````
