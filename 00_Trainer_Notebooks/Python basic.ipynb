```python
#Dictionary...!

groc = {'Rice':10,'Wheat':10,'Milk':100,'Rice':80,'Oil':180}
print(groc)

print(groc.items()) #for printing all items in a dictionary

print(groc.keys())  #for printing all keys 

print(groc.values()) #for printing all values

print(groc['Milk'])  # for access a value

groc['fruit'] = 200  # adding an item
groc.pop('Oil')     #removing an item
print(groc) #before update

groc.update({'Oil':180}) # for updating an item
print(groc) #after update

if 'Rice' in groc:  #for checking item is present 
    print("yes,rice is there")

if 'Biryani' not in groc:  #for checking item is not present 
    print("no,biryani is not there")
```
