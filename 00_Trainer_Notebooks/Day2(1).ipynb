## 2. Identifiers (Naming Rules)

An **identifier** is any name you give to a variable, function, class, etc.

### Rules:

| Rule | Valid | Invalid |
|------|-------|---------|
| Start with letter or `_` | `name`, `_age` | `1name` |
| Can have letters, digits, `_` | `my_var2` | `my-var` |
| No special characters | `total` | `$alary`, `Full Name` |
| Cannot be a keyword | `score` | `class`, `if`, `for` |
| Case-sensitive | `Name` ≠ `name` | — |

## 3. Variables

A **variable** is a container that stores a value. Python is **dynamically typed** — no need to declare the type!

```python
x = 10         # Python auto-detects: integer
x = "hello"    # Now x is a string (type changed!)
```

**Analogy:** A variable is like a **label on a box** — you can stick the label on any box (any type of data).

### Key Points:
- Assignment uses `=` (not `==`)
- A variable can change its type anytime
- Python figures out the type automatically

````python
# --- Variables & Dynamic Typing ---

x = 10
print(x, type(x))     # int

x = "Hello"
print(x, type(x))     # str (same variable, different type!)

x = 3.14
print(x, type(x))     # float

# Multiple assignment
a, b, c = 10, 20, 30
print(a, b, c)
````

## 5. id() Function — Memory Address

`id()` returns the **memory address** of an object. This helps understand how Python stores variables.

**Key insight:**
- Same value → may share same memory (for small ints & strings)
- Different value → different memory address
- This concept is important for understanding **mutable vs immutable** (next topic)

````python
# --- id() Function ---

a = 10
c = 10
b = 20

print(f"a = {a}, id = {id(a)}")
print(f"c = {c}, id = {id(c)}")
print(f"b = {b}, id = {id(b)}")

# a and c have SAME value → SAME memory address!
print(f"\na and c same id? {id(a) == id(c)}")  # True
print(f"a and b same id? {id(a) == id(b)}")    # False
````
